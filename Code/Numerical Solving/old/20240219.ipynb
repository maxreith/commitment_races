```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # for 3D plotting

# Define the functions for p and q:
def p(b, c):
    # Avoid division by zero: assume b, c chosen so that 1 - b - c > 0.
    return - (b**2 * c) / ((1 - b - c) * (1 - b - c + 2*b*c))

def q(b, c, p_val):
    # b-1 is always < 0 since b<0 so no risk of division by zero here.
    return (c * p_val - 1) / (b - 1)

# Create a grid for b and c.
# Here we choose b in [-2, -0.1] (since b<0) and c in [-2, 0.99] (since c<1).
b_vals = np.linspace(-2, -0.1, 400)
c_vals = np.linspace(-2, 0.99, 400)
B, C = np.meshgrid(b_vals, c_vals)

# Compute p and q over the grid.
P = p(B, C)
Q = q(B, C, P)

# Define the regions where p and q are in [0,1].
validP = (P >= 0) & (P <= 1)
validQ = (Q >= 0) & (Q <= 1)
# Determine where both p and q are in [0,1]:
validBoth = validP & validQ
```

# Plots

```python
# Create three 2D plots in one row.
fig, axs = plt.subplots(1, 3, figsize=(18, 6))

# Plot 1: Region where q is valid (i.e. q in [0,1])
cf1 = axs[0].contourf(B, C, validQ.astype(int), levels=[-0.5, 0.5, 1.5], colors=['red','green'], alpha=0.5)
axs[0].set_xlabel('b')
axs[0].set_ylabel('c')
axs[0].set_title('Region where q in [0,1]')

# Plot 2: Region where p is valid (i.e. p in [0,1])
cf2 = axs[1].contourf(B, C, validP.astype(int), levels=[-0.5, 0.5, 1.5], colors=['red','green'], alpha=0.5)
axs[1].set_xlabel('b')
axs[1].set_ylabel('c')
axs[1].set_title('Region where p in [0,1]')

# Plot 3: Region where both p and q are valid (both in [0,1])
cf3 = axs[2].contourf(B, C, validBoth.astype(int), levels=[-0.5, 0.5, 1.5], colors=['red','green'], alpha=0.5)
axs[2].set_xlabel('b')
axs[2].set_ylabel('c')
axs[2].set_title('Region where both p and q in [0,1]')

plt.tight_layout()
plt.show()
```
