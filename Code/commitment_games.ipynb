# Commitment Games Analysis

This notebook uses the `compute_base_and_commitment_equilibria` function to transform base games into commitment games and analyze their equilibria. Each game is analyzed by comparing the equilibria of the original base game with those of the corresponding commitment game.

```python
import numpy as np
import sys
sys.path.append('/Users/maxreith/GitHub/commitment_races/Code')
from transform_game import compute_base_and_commitment_equilibria, print_equilibria
```

### Game of Chicken

```python
payoff_player1 = np.array([[0, 2],
                           [0.5, 1]])

equilibria, games = compute_base_and_commitment_equilibria(payoff_player1, filter_for_symmetric=True)
print_equilibria(equilibria, games)
```

### Battle of Sexes

```python
payoff_player1 = np.array([[0, 2],
                           [1.5, 1]])

equilibria, games = compute_base_and_commitment_equilibria(payoff_player1, filter_for_symmetric=True)
```

### Stag Hunt
Note that there should actually be arbitrary many MSNE where players mix between $A_2$ and $W_{A_2}$. All of these yield payoff $1$.

```python
payoff_player1 = np.array([[0, 0.5],
                           [-1, 1]])
equilibria, games = compute_base_and_commitment_equilibria(payoff_player1, filter_for_symmetric=True)
print_equilibria(equilibria, games)
print(games["commitment_game"])
```
