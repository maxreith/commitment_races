# Solving some games

```python
import nashpy as nash
import numpy as np
```

```python
from compy.transform_game import transform_game
```

# Solve a Game and its Commitment Version

```python
# Row player payoffs (A, B, C vs X, Y, Z)
P1 = np.array([
    [3, -1, -1],  # A
    [1, -2, 1],   # B
    [1, 2, -2]    # C
])

# Column player payoffs
P2 = np.array([
    [1, -3, 0],   # A
    [2, 1, 0],    # B
    [2, 3, 3]     # C
])

# Create the game object
game = nash.Game(P1, P2)

print(game)
```

```python
# Transform the game using compy
transformed_game = transform_game(game)

# Orignal Game
print("Original game payoff matrices:")
print(game)

# Print equilibria of the original game
print("\nEquilibria of the original game:")
for eq in game.support_enumeration():
    print(eq)
    print("Payoffs:", game[eq])

# Transformed Game
print("\n\nTransformed Commitment Game payoff matrices:")
print(transformed_game)

# Print equilibria of the transformed game
print("\nEquilibria of the transformed game:")
for eq in transformed_game.support_enumeration():
    print(eq)
    print("Payoffs:", transformed_game[eq])
```

# Cournout Game

```python
P1 = np.array([
    [0, 1/12, 1/8],
    [1/24, 1/9, 5/36],
    [1/16, 5/48, 1/8]
])

game = nash.Game(P1, P1.T)
```

```python
# Solve the game and print equilibria
print("Equilibria of the game:")
for eq in game.support_enumeration():
    print(eq)
    print("Payoffs:", game[eq])
```

```python
game_transformed = transform_game(game)

print(game_transformed)

print()
print("Mixed equilibria of the game:")
tol = 1e-12
found = False
for eq in game_transformed.support_enumeration():
    x, y = eq
    if (np.count_nonzero(x > tol) > 1) or (np.count_nonzero(y > tol) > 1):
        found = True
        print(eq)
        payoffs = tuple(np.round(game_transformed[eq], 12))
        print("Payoffs:", payoffs, "\n")
if not found:
    print("None")
```

While I am getting NEOs here, might be because I have a simplified version of the Cournot game here where the quantites are discrete rather than continuous.

However, this example shows one other interesting thing. You can get mixed NEOs in games that do not have MSNE!
